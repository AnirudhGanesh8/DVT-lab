```python
import pandas as pd
import sqlite3
```

```python
# Read the dataset

df = pd.read_csv("all_space_mission_launches.csv")

print("Dataset Imported Successfully!")
```

```python
print(df.columns.tolist())         #Display all column names
```

```python
print("Duplicate Rows :", df.duplicated().sum())       #Check no. of Duplicate rows
```

```python
df.sample(5, random_state=42)
```

```python
df.to_excel("space_missions.xlsx", index=False)  #Import data from excel

print("Excel File Created Successfully!")
```

```python
excel_df = pd.read_excel("space_missions.xlsx")

excel_df.head()
```

```python
conn = sqlite3.connect("space_missions.db")    #Import Data from SQL Database

df.to_sql("missions", conn, if_exists="replace", index=False)

print("Data Stored in SQL Database")
```

```python
query = "SELECT * FROM missions"     #Read from SQL

sql_df = pd.read_sql(query, conn)

sql_df.head()
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"   #Import Data from Web

web_df = pd.read_csv(url)

web_df.head()
```

```python
df.to_csv("space_mission_export.csv", index=False)      #Export Dataset to CSV

print("CSV File Exported Successfully!")
```

```python
df.to_excel("space_mission_export.xlsx", index=False)    #Export Dataset to Excel

print("Excel File Exported Successfully!")
```

```python
df.to_json("space_mission_export.json", orient="records")     #Export Dataset to JSON

print("JSON File Exported Successfully!")
```

```python
import os              #Verify exported Files

files = [
    "space_mission_export.csv",
    "space_mission_export.xlsx",
    "space_mission_export.json"
]

for file in files:
    if os.path.exists(file):
        print(file, "Created Successfully")
    else:
        print(file, "Not Found")
```
